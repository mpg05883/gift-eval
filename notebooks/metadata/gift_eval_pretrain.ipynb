# GIFT-Eval Pretrain Metadata

Load the metadata that was collected and compare it to the values in Table 14 of the [GIFT-Eval paper](https://arxiv.org/pdf/2410.).

```python
import pandas as pd
from pathlib import Path

pd.set_option("display.max_rows", None)

# corpus = "gift_eval"
corpus = "gift_eval_pretrain"
path = Path(f"../../data/metadata/{corpus}.csv")
df = pd.read_csv(path)
df.style.format("{:,}", subset=df.select_dtypes("number").columns)


print(f"Shape: {df.shape}")
df.head()
# display(df)
```

The `sum_series_length` column contains the observations across *all channels*. However, we want to treat the channels across each timestamp as one observation. Divide `sum_series_length` by `target_dim` to count the total number of observations regardless of channels.

```python
mask = df["target_dim"] > 1
df.loc[mask, "sum_series_length"] = (
    df.loc[mask, "sum_series_length"] / df.loc[mask, "target_dim"]
).astype(int)
display(df)
```
